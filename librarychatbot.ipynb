```python
# ── CELL 3: TRAIN TF-IDF MODEL & ADVANCED NLP DISPATCHER ───
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import re

questions = [pair[0] for pair in training_data]
answers   = [pair[1] for pair in training_data]

# Vectorize questions using TF-IDF (1-2 ngrams)
vectorizer = TfidfVectorizer(ngram_range=(1, 2), sublinear_tf=True)
question_vectors = vectorizer.fit_transform(questions)

def get_colab_reply(user_message, threshold=0.18):
    text = (user_message or '').strip()
    if not text:
        return "Hello! Please enter a book title, author, category, room inquiry, or question."
    
    lower = text.lower()
    
    # 1. Exact Book ID Match (B01 - B80)
    id_match = re.search(r'\bB0?([1-9]|[1-7][0-9]|80)\b', text, re.IGNORECASE)
    if id_match:
        bid = "B" + str(id_match.group(1)).zfill(2)
        book = next((b for b in BOOKS_DATA if b['id'].upper() == bid.upper()), None)
        if book:
            return (
                f"📖 <b>Book ID: {book['id']}</b> — <b>{book['title']}</b><br>"
                f"👤 Author: {book['author']}<br>"
                f"📂 Category: {book['category']}<br>"
                f"📍 Location: <b>Shelf {book['shelf']}, Row {book['row']}</b><br>"
                f"📦 Total Copies: {book['copies']} | Available: {book['copies']}<br>"
                f"🏷️ Keywords: <i>{book['keywords']}</i>"
            )
            
    # 2. Room ID Match (R01 - R06)
    room_match = re.search(r'\b(R0[1-6])\b', text, re.IGNORECASE)
    if room_match:
        rid = room_match.group(1).upper()
        room = next((r for r in ROOMS_DATA if r['id'] == rid), None)
        if room:
            return (
                f"🏢 <b>Room {room['id']}: {room['name']}</b> ({room['type']})<br>"
                f"📍 Location: {room['floor']}<br>"
                f"👥 Capacity: <b>{room['capacity']}</b><br>"
                f"🛠️ Equipment: {room['equipment']}<br>"
                f"🎯 Best For: <i>{room['suitable_for']}</i><br><br>"
                f"To book this room, type: 'Book {room['id']} for [Your Student ID]'"
            )

    # 3. Location Queries ("where is ...", "location of ...")
    if any(k in lower for k in ["where is", "location of", "shelf of", "where can i find"]):
        clean_query = re.sub(r'^(where is|location of|shelf of|where can i find)\s+', '', lower).strip(' ?.')
        for b in BOOKS_DATA:
            if clean_query in b['title'].lower() or clean_query in b['id'].lower():
                return f"📍 Location for <b>'{b['title']}'</b> ({b['id']}): <b>Shelf {b['shelf']}, Row {b['row']}</b> ({b['category']})."

    # 4. TF-IDF Cosine Similarity Matching
    user_vec = vectorizer.transform([lower])
    similarities = cosine_similarity(user_vec, question_vectors)[0]
    best_index = int(np.argmax(similarities))
    best_score = float(similarities[best_index])
    
    if best_score >= threshold:
        return answers[best_index]
        
    # 5. Keyword Substring Fallback for Titles and Authors
    matching_books = []
    for b in BOOKS_DATA:
        if lower in b['title'].lower() or lower in b['author'].lower() or lower in b['keywords'].lower():
            matching_books.append(b)
            if len(matching_books) >= 3:
                break
                
    if matching_books:
        res = "🔍 I found matching books in the catalogue:<br><br>"
        for b in matching_books:
            res += f"• <b>{b['id']}: {b['title']}</b> by {b['author']} (Shelf {b['shelf']}, Row {b['row']})<br>"
        return res
        
    # 6. General Helpful Fallback
    return (
        "I couldn't find a direct match for your request. "
        "Here are some things you can try:<br>"
        "• Search by Book ID (e.g. <code>B01</code> or <code>B72</code>)<br>"
        "• Inquire about study rooms (e.g. <code>book room</code>, <code>R01</code>)<br>"
        "• Ask for library hours (e.g. <code>library hours</code>, <code>weekend hours</code>)<br>"
        "• Check overdue fines (e.g. <code>overdue fines</code>, <code>late fee</code>)<br>"
        "• Browse categories (e.g. <code>categories</code>, <code>Fiction books</code>)."
    )

# Quick sanity checks
print('🧪 Testing NLP Engine:')
print('1. B01 Test:', get_colab_reply('B01')[:60] + '...')
print('2. Room Test:', get_colab_reply('R01')[:60] + '...')
print('3. Hours Test:', get_colab_reply('library hours')[:60] + '...')
print('4. Fines Test:', get_colab_reply('overdue fines')[:60] + '...')
print('✅ NLP Engine is trained and ready!')
```

```python
# ── CELL 4: START FLASK REST API SERVER ────────────────────
from flask import Flask, request, jsonify, make_response
from flask_cors import CORS
import threading

app = Flask(__name__)
# Enable CORS for all routes and origins
CORS(app, resources={r"/*": {"origins": "*"}}, supports_credentials=True)

@app.after_request
def add_cors_headers(response):
    response.headers['Access-Control-Allow-Origin'] = '*'
    response.headers['Access-Control-Allow-Methods'] = 'GET, POST, OPTIONS'
    response.headers['Access-Control-Allow-Headers'] = 'Content-Type, Authorization, ngrok-skip-browser-warning, Bypass-Tunnel-Reminder'
    return response

# Ping endpoint for health and connection testing
@app.route('/ping', methods=['GET'])
def ping():
    return jsonify({
        'status': 'online',
        'server': 'APU Library AI Colab Backend',
        'books_loaded': len(BOOKS_DATA),
        'rooms_loaded': len(ROOMS_DATA),
        'training_pairs': len(training_data)
    })

# Main /chat endpoint
@app.route('/chat', methods=['POST', 'OPTIONS'])
def chat():
    if request.method == 'OPTIONS':
        return make_response(jsonify({'status': 'ok'}), 200)
        
    data = request.get_json(silent=True) or {}
    message = data.get('message', '')
    reply = get_colab_reply(message)
    return jsonify({
        'reply': reply,
        'source': 'colab_ai',
        'status': 'success'
    })

# Optional endpoints
@app.route('/books', methods=['GET'])
def get_books():
    return jsonify({'books': BOOKS_DATA, 'count': len(BOOKS_DATA)})

@app.route('/rooms', methods=['GET'])
def get_rooms():
    return jsonify({'rooms': ROOMS_DATA, 'count': len(ROOMS_DATA)})

# Start Flask in background thread
threading.Thread(target=lambda: app.run(host='0.0.0.0', port=5000, use_reloader=False), daemon=True).start()
print('✅ Flask REST API server is running on port 5000 in background thread!')
```
